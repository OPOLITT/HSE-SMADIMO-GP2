```python
from dotenv import load_dotenv
import pandas as pd
import requests
import os
```

```python
load_dotenv()

API_KEY = os.getenv('API_KEY')
```

```python
search_url = "https://api.themoviedb.org/3/search/movie"

params = {
    "api_key": API_KEY,
    "query": "White Chicks",
    "language": "en-US"
}

headers = {
    "accept": "application/json"
}

response = requests.get(search_url, params=params, headers=headers)

response.json()
```
